MAP API call test

```python
import requests

def test_geocoding(api_key, address):
    url = f"https://maps.googleapis.com/maps/api/geocode/json?address={address}&key={api_key}"
    
    response = requests.get(url)
    data = response.json()
    
    if data['status'] == 'OK':
        # Get the first result's coordinates
        location = data['results'][0]['geometry']['location']
        print(f"✅ Success! {address}")
        print(f"Latitude: {location['lat']}")
        print(f"Longitude: {location['lng']}")
    else:
        print(f"❌ Error: {data['status']}")
        print(f"Message: {data.get('error_message', 'No message')}")

# Replace with your key
MY_KEY = "YOUR_GOOGLE_MAPS_API_KEY" 
test_geocoding(MY_KEY, "Near Balajinagar, Satara Road, Pune, Maharashtra, India")
```

MAP key test for 10 entries

```python
import pandas as pd
import requests
import time

# === CONFIG ===
API_KEY = "YOUR_GOOGLE_MAPS_API_KEY"
INPUT_FILE = "data/processed/accidents_processed.csv"

# === Load Data ===
df = pd.read_csv(INPUT_FILE)

# Take only first 10 rows
sample_df = df.head(10)

print(f"Testing {len(sample_df)} locations...\n")

def geocode_google(address, api_key):
    url = "https://maps.googleapis.com/maps/api/geocode/json"
    params = {
        "address": address,
        "key": api_key
    }

    response = requests.get(url, params=params)
    data = response.json()

    if data["status"] == "OK":
        location = data["results"][0]["geometry"]["location"]
        return location["lat"], location["lng"]
    else:
        return None, None


# === Test Loop ===
for idx, row in sample_df.iterrows():
    address = row["location"]

    lat, lon = geocode_google(address, API_KEY)

    print("=" * 70)
    print(f"📝 Title: {row['title']}")
    print(f"📍 Location Text: {address}")

    if lat is not None:
        print(f"✅ Latitude: {lat}")
        print(f"✅ Longitude: {lon}")
    else:
        print("❌ Geocoding failed")

    time.sleep(0.2)  # small delay to be polite
```
